## Configuration

Paths are resolved from the repository root and can be overridden with the
`SMELL_*` environment variables documented in the README.

````python
from smell_space.config import (
    DATASET_PATH,
    EMBEDDINGS_PATH,
    METADATA_PATH,
    MODEL_DIR,
    PROJECT_ROOT,
)

print(f"Repository root: {PROJECT_ROOT}")
print(f"Dataset: {DATASET_PATH}")
print(f"Model checkpoints: {MODEL_DIR}")
print(f"Embeddings: {EMBEDDINGS_PATH}")
print(f"Metadata: {METADATA_PATH}")
````

## Inspect generated artifacts

These files are produced by `python scripts/extract_embeddings.py` after a
model has been trained.

````python
import numpy as np
import pandas as pd

if not EMBEDDINGS_PATH.exists() or not METADATA_PATH.exists():
    raise FileNotFoundError(
        "Run scripts/extract_embeddings.py to generate the demo artifacts."
    )

embeddings = np.load(EMBEDDINGS_PATH)
metadata = pd.read_csv(METADATA_PATH)
print(f"Embeddings: {embeddings.shape} {embeddings.dtype}")
print(f"Metadata rows: {len(metadata):,}")
metadata.head()
````

## UMAP map

The same projection is cached by the extraction script; this is only a
notebook-side visualization.

````python
import plotly.express as px

fig = px.scatter(
    metadata,
    x="umap_x",
    y="umap_y",
    color="dominant_family",
    hover_name="name",
    hover_data=["smiles", "active_descriptors"],
    title="OpenPOM latent odor space",
    height=600,
)
fig.show()
````
